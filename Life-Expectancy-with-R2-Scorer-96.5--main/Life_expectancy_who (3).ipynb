# **Gradient Boosting Regressor**

Gradient Boosting is a machine learning technique that builds models sequentially, where each new model attempts to correct the errors of the previous models. The models are typically decision trees, and the goal is to minimize the loss function by combining the predictions of each model.

# **XGB Regressor (XGBoost)**

XGBoost stands for Extreme Gradient Boosting. It is an optimized implementation of gradient boosting, designed to be highly efficient and flexible. XGBoost builds upon the gradient boosting algorithm by adding regularization, which helps prevent overfitting, and uses advanced techniques to handle missing data and to make use of sparsity in the data.

##  ***4. Data preprocessing :***

<a name='2.1'></a>
## 4.1 data processing

```python
train_input_data = data_train
```

<a name='2.2'></a>
## 4.2 handling missing values

**apply simple imputer in numrical data :**

```python

from sklearn.impute import SimpleImputer

numrical_train_data=train_input_data.drop(['Country','Status'],axis=1)

imputer=SimpleImputer(strategy='mean')

train_data_without_missingValues=imputer.fit_transform(numrical_train_data)
```

**convert result to dataframe :**

```python


numrical_train_data_imputed=pd.DataFrame(train_data_without_missingValues,columns=numrical_train_data.columns,index=numrical_train_data.index)

numrical_train_data_imputed.isna().sum()
```

**add this result to train_input_data :**

```python

train_input_data[numrical_train_data.columns] = numrical_train_data_imputed

train_input_data.isnull().sum()
```

<a name='4.3'></a>
## 2.3 Encoding categorical features

**encode categorical column with ordinary encoder :**

```python

from sklearn.preprocessing import OrdinalEncoder

cat_col=train_input_data[['Status','Country']]

encoder=OrdinalEncoder()

encoded_col=encoder.fit_transform(cat_col)
```

**convert result to data frame :**

```python

encoded_col_df=pd.DataFrame(encoded_col,columns=cat_col.columns,index=cat_col.index)
```

**add this col to train_input_data:**

```python

train_input_data[['Status','Country']]=encoded_col_df

train_input_data.head()
```

<a name='2.4'></a>
## 4.4 Scaling Numerical features

**Scaling Numerical features using snaderizaion :**

```python

from sklearn.preprocessing import StandardScaler

scaler=StandardScaler()

scalered_train_data=scaler.fit_transform(train_input_data)

scalered_train_data
```

**convert result to data frame :**

```python

scalered_train_data_df=pd.DataFrame(scalered_train_data,columns=train_input_data.columns)
```

**add this to train_input_data :**

```python

train_input_data=scalered_train_data_df

train_input_data.head()
```

<a name='5'></a>
##  ***5. Training and Evaluating models :***

**Divde data to train and labels :**

```python
x=train_input_data.drop('Life expectancy ',axis=1)
y=train_input_data['Life expectancy ']
```

**Split data to Train and Test :**

```python
from sklearn.model_selection import train_test_split
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size = 0.2, random_state = 30)
```

**import all nedeed models**

```python
from sklearn.linear_model import LinearRegression
from sklearn.ensemble import RandomForestRegressor, GradientBoostingRegressor
from xgboost import XGBRegressor
from sklearn.metrics import mean_squared_error, r2_score
```

**Train And Evalute Models :**

```python
models = {
    'Random Forest': RandomForestRegressor(random_state=42),
    'GradientBoost Regressor': GradientBoostingRegressor(random_state=42),
    'XGB Regressor': XGBRegressor(),
    'Linear Regression': LinearRegression()
}

results = []

for model_name, model in models.items():

    model.fit(x_train, y_train)

    prediction = model.predict(x_test)

    rmse = mean_squared_error(y_test, prediction, squared=False)
    r2 = r2_score(y_test, prediction)

    results.append({'Model': model_name, 'RMSE': rmse, 'R2 Score': r2})

results_df = pd.DataFrame(results)
```

```python
results_df=results_df.sort_values("R2 Score", ascending = False)
results_df
```

**Compute Cros validation for the best model:**

```python
from sklearn.model_selection import cross_val_score, KFold

best_model = XGBRegressor()

kf = KFold(n_splits=20, shuffle=True, random_state=42)

cross_val = cross_val_score(best_model,x, y, cv= kf, scoring='r2')
cross_val
```

```python
plt.plot(cross_val)
plt.xlabel('Fold')
plt.ylabel('Score')
plt.title("Cross Validation scores")
```

```python
mean=cross_val.mean()
mean
```

**Get the final cross validation percentage**

```python
print("final Accuracy percentage: ", mean * 100, "%")
```
